# 03 Model Comparison

Review selected model performance, ranking, confusion matrix, ROC curve, and precision-recall curve.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from src.utils import REPORTS_DIR, artifact_paths, load_config, load_json

sns.set_theme(style='whitegrid')
paths = artifact_paths(load_config())
metrics = load_json(paths['metrics_path'])
```

```python
comparison = pd.DataFrame(metrics['comparison'])
comparison[['model', 'accuracy', 'precision', 'recall', 'f1', 'roc_auc', 'cv_mean', 'cv_std']]
```

```python
plt.figure(figsize=(9, 5))
ordered = comparison.sort_values('roc_auc', ascending=False)
sns.barplot(data=ordered, x='roc_auc', y='model', color='#1167b1')
plt.xlim(0, 1)
plt.title('Model Ranking by ROC-AUC')
plt.tight_layout()
plt.savefig(REPORTS_DIR / 'notebook_model_ranking.png', dpi=180)
plt.show()
```

```python
matrix = metrics['confusion_matrix']
plt.figure(figsize=(5, 4))
sns.heatmap(matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['Predicted Benign', 'Predicted Malignant'], yticklabels=['Actual Benign', 'Actual Malignant'])
plt.title('Confusion Matrix')
plt.tight_layout()
plt.show()
```

```python
roc = metrics['curves']['roc']
plt.figure(figsize=(7, 5))
plt.plot(roc['fpr'], roc['tpr'], color='#1167b1', linewidth=2, label=f"ROC-AUC = {metrics['test_metrics']['roc_auc']:.3f}")
plt.plot([0, 1], [0, 1], linestyle='--', color='#98a2b3')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
pr = metrics['curves']['precision_recall']
plt.figure(figsize=(7, 5))
plt.plot(pr['recall'], pr['precision'], color='#0f766e', linewidth=2)
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.tight_layout()
plt.show()
```
